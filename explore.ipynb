```python
def get_url(source_title, version):
    return f"https://simple.wikipedia.org/w/index.php?title={source_title}&oldid={version}"
```

```python
def get_file_name(source_ID, version):
    return f"{source_ID}_{version}.html"
```

```python
url_df = pd.DataFrame()

url_df['title'] = scored_simple_5['source_title']
url_df['first_file_name'] = scored_simple_5[['source_ID', 'first_version']].apply(lambda x: get_file_name(x['source_ID'], x['first_version']), axis=1)
url_df['second_file_name'] = scored_simple_5[['source_ID', 'second_version']].apply(lambda x: get_file_name(x['source_ID'], x['second_version']), axis=1)
url_df['first_version_url'] = scored_simple_5[['source_title', 'first_version']].apply(lambda x: get_url(x['source_title'], x['first_version']), axis=1)
url_df['second_version_url'] = scored_simple_5[['source_title', 'second_version']].apply(lambda x: get_url(x['source_title'], x['second_version']), axis=1)

display(url_df)
```

```python
for i, first_verison_url in enumerate(url_df['first_version_url']):
    print(i)
    print(url_df['first_file_name'][i])
    print(first_verison_url)
```
